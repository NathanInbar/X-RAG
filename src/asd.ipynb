```python
dataset_path = CWD / "../datasets/MINE/"
if not dataset_path.is_dir():
    print("cant find dir")

import json
from utils import Tokenizer
_tokenizer = Tokenizer()

tk_lens = []
for i,p in enumerate(dataset_path.iterdir()):
    with open(p, "r") as fp:
        data = json.load(fp)
        tks = _tokenizer.encode(data['essay'])
        tk_lens.append(len(tks))
    if i == 0:
        print(f"{p.stem} : {len(tks)}")
    if len(tks) > 1200:
        print(p.stem)

from statistics import mean, median
print(f"mean MINE tokens:  {mean(tk_lens)}")
print(f"median MINE tokens: {median(tk_lens)}")
print(f"min toks: { min(tk_lens)}")
```
